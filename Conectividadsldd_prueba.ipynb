# 1.1 Conectividad
* No es una propiedad intrinseca del paisaje, sino que es considerada como **“una relación entre un paisaje, tecnologías de movimiento y un sistema social con lógicas culturales e históricas propias”.**

* Tayor et.al (1993:571) la definen como **“el grado en que el paisaje facilita o impide el movimiento entre teselas de recursos”**
* Conectividad: Estructural versus funcional
Distinción que proviene de la Ecología del Paisaje (Taylor et.al, 1993) donde propone:

  a. Conectividad Estructural: describe la configuración del medio biofísico del espacio: continuidad de habitats, presencia de corredores, barreras topográficas, etc.


  b. Conectividad funcional: incorpora la respuesta de organismos/agentes a esa estructura, dependiendo de su capacdad de movimiento, percepción y motivación.


* Conectividad del Paisaje en arqueología: Howey (2007:1831) traslada esta definición al aplicar los métodos de superficie de costo multicriterio, señalando que “la conectividad funcional puede utilizarse para inferir paisajes sociales del pasado”.
**texto en negrita**

* Crítica a la utilización del concepto de “conectividad funcional” en arqueología: equiparan y reducen sólo a una función de minimización de costos energéticos, ya que funcionalidad incluye aspectos biomecanicos, cognitivos, políticos, simbólicos, religiosos, econoḿicos, etc.

# 1.2 Conectividad como movilidad, intercambio, interacción y/o control

* Usos del término “conectividad” en arqueología:
  - Movilidad residencial o logística de cazadores recolectores

  - Redes de intercambio de bienes que mapean esferas de interacción

  - Peregrinaje y paisajes rituales, donde el movimiento en si mismo es una experiencia cargada de significado propio.
  
  - Control territorial y vialidad imperial, donde se concibe la conectividad como instrumento de integración política y fiscal (el caso del Qhapaq Ñan).

* Cada escenario implica motivaciones y modos distintos de tránsito que determinarán la ruta. No es lo mismo el traslado de obsidiana en llama que la peregrinacion ceremonial a un punto especifico. Se espera que en el primer caso se priorice aspectos como la seguridad, forraje y eficiencia energética, mientras que en el segundo caso se espera que prevalezcan aspectos como la visibilidad con elementos del paisaje, la intervisibilidad entre sitios/cumbres, privilegio del paso por sobre el coste energetico, etc.

* Tim Ingold (2007): señala la diferencia entre:
  - **“transporte”** (moverse a través del paisaje como si fuera una superficie abstracta a cruzar con el menor costo) y
  - **“wayfaring”** (habitar el movimiento donde el trayecto es una parte constitutiva de la experiencia y el conocimiento). El “wayfiring” admite desvíos, redundancias, lecturas de señales dentro de los territorios, etc.

* Snead (2009: 42-43): considera que **“los senderos no son simplemente lineas de movimiento eficiente, sino estructuras que reflejan relaciones sociales, memoria y negociación”**.

* Acuto (2010): poblaciones incaicas resignificaron rutas pre-existentes para imponer su propio orden espacial del mundo. Implica no solo la capacidad de movimiento en si mismo, sino la capacidad de control sobre las rutas.

# 1.3 La trampa del “camino único” y el “agente omnisciente”

* LCP asume al viajero como un agente omnisciente: conocimiento perfecto del terreno que puede evaluar todos los costos, permitiendo la elección de la ruta la función predefinida. Llobera (2000) llamó a esta caracterización como *“piloto de movimiento omnisciente".*

Sin embargo, se reconoce actualmente que actores reales tienen conocimiento parcial, sendas heredadas e información visual local. **El movimiento humano se caracteriza por la “redundancia”, variabilidad entre individuos y grupos, así como por la consideración de múltiples rutas alternativas viables. No se caracteriza por ser determinista**.

* Howey (2011:2524) considera que *“el LCP modela el movimiento como una optimización singular, lo cual asume un conocimiento perfecto y a un viajero estandar. Esto no refleja la complejidad del movimiento humano real, en el cual múltiples caminos pueden ser utilizados simultaneamente”.*

* Fábrega-Alvarez (2006: 7-8): modelamiento “sin destino”, es decir, desde un origen pero con múltiples direcciones que emerjan desde la topografía. Produce corredores difusos más cercanos al registro etnografico de movilidad de pastores. Es decir, flujos multiples, no un vector único.

# 1.4 La trampa al confundir conectividad física con significado cultural
Existe una tendencia hacia cometer el error de asumir que alta conectividad física es equivalente a una alta interacción social.
* La accesibilidad física no es lo mismo que laLa accesibilidad puede tener restricciones del ambito biofísico o social (vedados por razones rituales, etnicas y/o politicas)

* Conectividad Arqueológica es socialmente mediada, por lo cual otras lineas de evidencia (presencia de bienes foraneos, estilos/tipologias de objetos, etc) pueden no ser coherentes con la cercanía/lejanía topográfica.

#1.6 LCP con simulación de Monte Carlo
La pregunta se traslada desde “cual es la ruta mas optima?” a “cual es la probabilidad de que una celda del paisaje forme parte de una ruta de bajo costo?”, generando como resultado un mapa de densidad de rutas o “corredor probabilistico”.

* Conover (2018:167): el analisis de MC permite tratar la fricción como variable aleatoria en lugar de considerarla como valor determinista, generando una familia de superficies de costo plausibles y evaluando la estabilidad de las rutas resultantes. Reconoce que se desconoce la función de coste real, la arqueología solo puede proponer rangos plausibles basados en datos empíricos (fisiológicos, ecológicos, etnográficos y/o experimentales) y ver que rutas emergen consistentemente a pesar de la incertidumbre.

* Los parámetros tipicamente sometidos a simulación de MC en estudios de conectividad arqueológica son:
  - Pendiente crítica y coeficientes de la función de coste
  - Fricción de coberturas del suelo
  - Resistencia de barreras
  - Direccionalidad isotrópica/anisotrópica

* Kantner (2004:326-328) demostró que las rutas inferidas cambiaban significativamente al perturbar pesos de las coberturas. Su conclusión fue que sin análisis de sensibilidad (como el analisis MC) los LCP no son reproducibles ni fiables.

# Definir extensión del área (utm zona 19s, similar a Maule)

y <- seq(y_min, y_max, length.out = n_celdas)

# Generar elevación con tendencia: aumenta hacia el este y norte (cordillera)
elevacion <- 1000 +
  (coords$x - x_min) / (x_max - x_min) * 1500 +  # gradiente este

  rnorm(nrow(coords), 0, 80)                     # ruido natural

# Convertir a raster

    lcp <- shortestPath(tr, o, d, output = "SpatialLines")
    lcp_list[[paste0(origenes$id[i], "_", destinos$id[j])]] <- lcp
  }
}

# 4.2 Función para calcular área visible desde un punto
calcular_viewshed <- function(x, y, dem_raster, observer_height = 1.6, max_distance = 5000) {
  # Convertir coordenadas a objeto SpatialPoints
  punto <- vect(cbind(x, y), crs = crs(dem_raster))
  
  # Calcular viewshed (usando terra)
  # Primero, crear una capa de alturas del observador
  elev_punto <- extract(dem_raster, punto)[[1]]
  observer_alt <- elev_punto + observer_height
  
  # viewshed requiere un raster y una matriz de observadores
  # Para simplificar, usamos una función de visibilidad punto a punto
  # Nota: terra::viewshed() está en desarrollo, usamos raster::viewshed()
  dem_raster_base <- raster(dem_raster)
  punto_sp <- SpatialPoints(cbind(x, y), proj4string = CRS(crs(dem_raster)))
  
  # Calcular viewshed (puede ser lento en DEM grandes)
  # Por simplicidad, usamos un área de ejemplo reducida
  dem_recortado <- crop(dem_raster_base, extent(x - max_distance, x + max_distance,
                                                y - max_distance, y + max_distance))
  v <- viewshed(dem_recortado, punto_sp, observer_height = observer_height,
                visib_height = 0, maxdist = max_distance)
  
  # Área visible (número de celdas visibles * área por celda)
  area_visible <- sum(values(v) == 1, na.rm = TRUE) * res(dem_recortado)[1]^2
  return(area_visible)
}

n_simulaciones <- 99
areas_aleatorias_arte <- list()

for(sim in 1:n_simulaciones) {
  # Puntos aleatorios alrededor de sitios con arte
  temp_arte <- st_sample(buff_arte, size = n_random) %>% st_coordinates()
  areas_temp_arte <- sapply(1:nrow(temp_arte), function(i) {
    calcular_viewshed(temp_arte[i, "X"], temp_arte[i, "Y"], dem)
  })
  areas_aleatorias_arte[[sim]] <- areas_temp_arte
  
  # Puntos aleatorios alrededor de sitios sin arte
  temp_sar <- st_sample(buff_sar, size = n_random) %>% st_coordinates()
  areas_temp_sar <- sapply(1:nrow(temp_sar), function(i) {
    calcular_viewshed(temp_sar[i, "X"], temp_sar[i, "Y"], dem)
  })
  areas_aleatorias_sar[[sim]] <- areas_temp_sar
  
  if(sim %% 20 == 0) cat("Simulación", sim, "completada\n")
}
